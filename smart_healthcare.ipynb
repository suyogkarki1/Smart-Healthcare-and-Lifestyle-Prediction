### 📝 Note on Dropping the 'Age' Feature

* **The Problem:** The initial iteration of the model scored an unrealistic `0.9996` PR-AUC because the `age` feature perfectly split the target classes (No heart disease under 70, heart disease concentrated over 53). This resulted in an inflated coefficient weight of `60.33`.
* **The Fix:** Dropped `age` to resolve data leakage and prevent the model from using a mathematical shortcut.
* **The Outcome:** The model now evaluates real clinical indicators (Diabetes, Stroke, BMI, Smoking) and achieves a highly realistic, generalizable **PR-AUC of 0.9892**.

### Statistic Testing

```python
from scipy.stats import ttest_ind
import numpy as np
import pandas as pd

ttest_results = []

for col in num_col:
    group_D = df[df[target] == 1][col].dropna()   # has heart disease
    group_N = df[df[target] == 0][col].dropna()   # no heart disease

    t_stat, p_value = ttest_ind(group_D, group_N, equal_var=False)

    mean_D, mean_N = group_D.mean(), group_N.mean()
    mean_diff = mean_D - mean_N
    pooled_std = np.sqrt((group_D.std()**2 + group_N.std()**2) / 2)
    cohen_d = mean_diff / pooled_std

    ttest_results.append({
        'Feature': col,
        'Mean_Disease': mean_D,
        'Mean_NoDisease': mean_N,
        'Mean_Difference': mean_diff,
        't_statistic': t_stat,
        'p_value': p_value,
        'Cohens_d': cohen_d,
        'Result': 'Significant' if p_value < 0.05 else 'Not Significant'
    })

ttest_results_df = pd.DataFrame(ttest_results).sort_values(by='p_value')
ttest_results_df
```

```python
from scipy.stats import chi2_contingency
import numpy as np
import pandas as pd

binary_cols = ['diabetes', 'stroke', 'chest_pain', 'smoking', 'alcohol', 'fatigue']

chi2_results = []
for col in binary_cols:
    contingency = pd.crosstab(df[col], df[target])
    chi2, p, dof, _ = chi2_contingency(contingency)

    # Cramér's V (effect size for categorical association)
    n = contingency.sum().sum()
    min_dim = min(contingency.shape) - 1
    cramers_v = np.sqrt(chi2 / (n * min_dim)) if min_dim > 0 else np.nan

    chi2_results.append({
        'Feature': col,
        'chi2': chi2,
        'dof': dof,
        'p_value': p,
        'Cramers_V': cramers_v,
        'Result': 'Significant' if p < 0.05 else 'Not Significant'
    })

chi2_results_df = pd.DataFrame(chi2_results).sort_values(by='p_value')
chi2_results_df
```

### Split dataset into train/test 

```python
from sklearn.model_selection import train_test_split


X_train_cal, X_test_cal, y_train_cal, y_test_cal = train_test_split(X, y_heart, test_size=0.2, random_state=42, stratify=y_heart) #for classification


```

### Pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

binary_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent'))
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, num_col),
        ('cat', categorical_transformer, cat_col),
        ('bin', binary_transformer, binary_cols),
    ]
)
```

```python
print(preprocessor.transformers)
```

### MODEL Selection

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (RandomForestClassifier, GradientBoostingClassifier, 
                              AdaBoostClassifier, ExtraTreesClassifier)
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score, 
                           roc_auc_score, roc_curve, confusion_matrix, classification_report,
                           mean_squared_error, mean_absolute_error, r2_score)
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_val_score
```

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier
from sklearn.svm import SVC
from sklearn.utils.class_weight import compute_sample_weight

models = {
    'LogisticRegression': LogisticRegression( max_iter=1000, class_weight='balanced'),
    'RandomForest': RandomForestClassifier( n_estimators=200, random_state=42, class_weight='balanced'),
    'GradientBoosting': GradientBoostingClassifier(n_estimators=200 , random_state=42),
    'AdaBoost': AdaBoostClassifier(n_estimators=200,random_state=42),
    'SVC': SVC( probability=True, class_weight='balanced', random_state=42)
}

no_class_weight_support = ['GradientBoosting', 'AdaBoost']

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    'accuracy': 'accuracy',
    'balanced_accuracy': 'balanced_accuracy',
    'f1': 'f1',
    'recall': 'recall',
    'precision': 'precision',
    'roc_auc': 'roc_auc',
    'pr_auc': 'average_precision'
}

cv_results = []

for name, model in models.items():
    pipe = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])

    fit_params = None
    if name in no_class_weight_support:
        sample_weight = compute_sample_weight(class_weight='balanced', y=y_train_cal )

        fit_params = {'classifier__sample_weight': sample_weight}

    scores = cross_validate(
        pipe,
        X_train_cal,
        y_train_cal,
        cv=cv,
        scoring=scoring,
        n_jobs=-1,
        params=fit_params
    )

    cv_results.append({
        'Model': name,
        'Accuracy': scores['test_accuracy'].mean(),
        'Balanced_Accuracy': scores['test_balanced_accuracy'].mean(),
        'Precision': scores['test_precision'].mean(),
        'Recall': scores['test_recall'].mean(),
        'F1_Score': scores['test_f1'].mean(),
        'ROC_AUC': scores['test_roc_auc'].mean(),
        'PR_AUC': scores['test_pr_auc'].mean()
    })

cv_results_df = pd.DataFrame(cv_results).sort_values(
    by='PR_AUC',
    ascending=False
)

cv_results_df
```

### Hyperparameter Tuning

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV

winner_name = 'LogisticRegression'

param_grids = {
    'LogisticRegression': [
        {
            # L2 regularization
            'classifier__solver': ['lbfgs'],
            'classifier__penalty': ['l2'],
            'classifier__C': [0.001, 0.01, 0.1, 1, 10, 100],
            'classifier__class_weight': [None, 'balanced']
        },
        {
            # L1 and L2 regularization
            'classifier__solver': ['liblinear'],
            'classifier__penalty': ['l1', 'l2'],
            'classifier__C': [0.001, 0.01, 0.1, 1, 10, 100],
            'classifier__class_weight': [None, 'balanced']
        },
        {
            # Elastic Net regularization
            'classifier__solver': ['saga'],
            'classifier__penalty': ['elasticnet'],
            'classifier__C': [0.001, 0.01, 0.1, 1, 10, 100],
            'classifier__l1_ratio': [0.25, 0.5, 0.75],
            'classifier__class_weight': [None, 'balanced']
        }
    ]
}

# param_grids = {
#     'RandomForestClassifier': {
#         'classifier__n_estimators': [200, 400],
#         'classifier__max_depth': [None, 10, 20],
#         'classifier__min_samples_split': [2, 5, 10],
#         'classifier__min_samples_leaf': [1, 2, 4],
#         'classifier__max_features': ['sqrt', 'log2'],
#     }
# }

winner_pipe = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(
        # max_iter=3000,
        class_weight='balanced',
        random_state=42
    ))
])

grid = GridSearchCV(
    winner_pipe,
    param_grids[winner_name],   
    cv=cv,
    scoring='average_precision',
    n_jobs=-1
)

grid.fit(X_train_cal, y_train_cal)

print("Best params:", grid.best_params_)
print("Best CV PR-AUC:", grid.best_score_)

best_model = grid.best_estimator_
best_classifier = best_model.named_steps['classifier']

print(best_classifier)
```

### scikit-learn uses an inverse relationship, meaning \(C = \frac{1}{\lambda}\)

```python
import numpy as np

# 1. Get the best estimator and the classifier step (Using your exact variables)
best_model = grid.best_estimator_
best_classifier = best_model.named_steps['classifier']

# 2. Extract feature names from your pipeline's preprocessor
feature_names = best_model.named_steps['preprocessor'].get_feature_names_out()

# 3. Extract the final coefficients and flatten them using [0]
coefficients = best_classifier.coef_[0]

# 4. Find features where the coefficient is exactly 0
zero_features = [name for name, coef in zip(feature_names, coefficients) if coef == 0]
active_features = [name for name, coef in zip(feature_names, coefficients) if coef != 0]

# 5. Print the results
print(f"❌ Dropped Features (Coefficient = 0): {len(zero_features)}")
for name in zero_features:
    print(f"  - {name}")

print(f"\n✅ Kept Features (Coefficient != 0): {len(active_features)}")
# Sort kept features by their absolute impact to see the most important ones first
sorted_active = sorted(zip(feature_names, coefficients), key=lambda x: abs(x[1]), reverse=True)
for name, coef in sorted_active:
    if coef != 0:
        print(f"  - {name}: {coef:.4f}")
```

### 📊 Understanding the Confusion Matrix

Here is a simple breakdown of how our model performed on the test data:

* **The Good Guesses (965 Total Correct):** 
  * **688 True Negatives:** People with **No Heart Disease** who were correctly predicted as having no heart disease.
  * **277 True Positives:** People with **Heart Disease** who were correctly predicted as having heart disease.

* **The Mistakes (35 Total Wrong):**
  * **12 False Positives:** People with **No Heart Disease** who were accidentally predicted to have heart disease.
  * **23 False Negatives:** People with **Heart Disease** who were accidentally missed and predicted to have no heart disease.

**Conclusion:** 
The model is highly reliable. It successfully catches over 92% of actual **Heart Disease** cases using only clinical indicators like diabetes, smoking, and BMI, while keeping wrong predictions very low.
